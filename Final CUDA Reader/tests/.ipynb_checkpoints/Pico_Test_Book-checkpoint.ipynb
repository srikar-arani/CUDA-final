```python
# manual script 
sc = """e
var c
var p
var f
var r
set_pgstat_chan 0
set_pgstat_mode 3
set_max_bandwidth 400
set_pot_range -1000m 0m
cell_on
set_e -600m
wait 200m
meas_loop_swv p c f r -600m 0m 10m 50m 100
	pck_start
	pck_add p
	pck_add c
	pck_add f    
	pck_add r
	pck_end
endloop
on_finished:
cell_off

"""
```

```python
res = pico.runScript(sc,parseValue=True)
```

```python
p = np.array([i[0] for i in res[1][0]])
c = np.array([i[1]*1e6 for i in res[1][0]])
f = np.array([i[2]*1e6 for i in res[1][0]])
r = np.array([i[3]*1e6 for i in res[1][0]])
```

```python
fig,ax = plt.subplots()
ax.plot(p,r,'.',label='Reverse')
ax.plot(p,f,'x',label='Forward')
ax.legend()
plt.show()
```
